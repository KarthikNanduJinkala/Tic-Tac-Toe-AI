```python
# Tic-Tac-Toe Board

board = [" " for _ in range(9)]

def display_board():
    print()
    print(board[0] + " | " + board[1] + " | " + board[2])
    print("--+---+--")
    print(board[3] + " | " + board[4] + " | " + board[5])
    print("--+---+--")
    print(board[6] + " | " + board[7] + " | " + board[8])
    print()

display_board()
```

```python
# Human Player Move

def human_move():
    while True:
        try:
            move = int(input("Choose a position (0-8): "))

            if move in range(9) and board[move] == " ":
                board[move] = "X"
                break
            else:
                print("Invalid move. Choose an empty position from 0-8.")

        except ValueError:
            print("Please enter a number from 0 to 8.")

display_board()
human_move()
display_board()
```

```python
import random

# AI Player Move

def ai_move():
    empty_positions = [i for i in range(9) if board[i] == " "]

    if empty_positions:
        move = random.choice(empty_positions)
        board[move] = "O"
        print("AI chose position:", move)

ai_move()
display_board()
```

```python
# Minimax Algorithm

def check_winner(board):
    winning_combinations = [
        (0, 1, 2),
        (3, 4, 5),
        (6, 7, 8),
        (0, 3, 6),
        (1, 4, 7),
        (2, 5, 8),
        (0, 4, 8),
        (2, 4, 6)
    ]

    for a, b, c in winning_combinations:
        if board[a] == board[b] == board[c] != " ":
            return board[a]

    if " " not in board:
        return "Tie"

    return None


def minimax(board, maximizing):
    result = check_winner(board)

    if result == "O":
        return 1
    elif result == "X":
        return -1
    elif result == "Tie":
        return 0

    if maximizing:
        best_score = -float("inf")

        for i in range(9):
            if board[i] == " ":
                board[i] = "O"
                score = minimax(board, False)
                board[i] = " "
                best_score = max(best_score, score)

        return best_score

    else:
        best_score = float("inf")

        for i in range(9):
            if board[i] == " ":
                board[i] = "X"
                score = minimax(board, True)
                board[i] = " "
                best_score = min(best_score, score)

        return best_score
```

```python
# AI Move using Minimax

def best_ai_move():
    best_score = -float("inf")
    best_move = None

    for i in range(9):
        if board[i] == " ":
            board[i] = "O"
            score = minimax(board, False)
            board[i] = " "

            if score > best_score:
                best_score = score
                best_move = i

    if best_move is not None:
        board[best_move] = "O"
        print("AI chose position:", best_move)
```

```python
# Complete Tic-Tac-Toe Game

board = [" " for _ in range(9)]

def play_game():
    print("Welcome to Tic-Tac-Toe AI!")
    print("You are X and the AI is O.")

    while True:
        display_board()

        # Human turn
        human_move()

        result = check_winner(board)
        if result is not None:
            display_board()
            if result == "X":
                print("🎉 You win!")
            else:
                print("It's a tie!")
            break

        # AI turn
        best_ai_move()

        result = check_winner(board)
        if result is not None:
            display_board()
            if result == "O":
                print("🤖 AI wins!")
            else:
                print("It's a tie!")
            break

play_game()
```
